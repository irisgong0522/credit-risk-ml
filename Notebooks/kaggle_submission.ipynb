```python
import pandas as pd
import joblib
from pathlib import Path

PROJECT_ROOT = Path("/Users/hongmo/Desktop/Gradient Boosting")

model_path = PROJECT_ROOT / "outputs" / "xgboost_refitted.joblib"

final_pipeline = joblib.load(model_path)

test_path = PROJECT_ROOT / "data" / "application_test.csv"

test_df = pd.read_csv(test_path)

print(test_df.shape)
display(test_df.head())
```

```python
test_ids = test_df["SK_ID_CURR"]

X_kaggle = test_df.drop(columns=["SK_ID_CURR"])
```

```python
test_probabilities = final_pipeline.predict_proba(
    X_kaggle
)[:, 1]
```

```python
submission = pd.DataFrame({
    "SK_ID_CURR": test_ids,
    "TARGET": test_probabilities
})

display(submission.head())

submission_path = (
    PROJECT_ROOT / "outputs" / "submission.csv"
)

submission.to_csv(submission_path, index=False)

print("Submission saved to:", submission_path)
```
