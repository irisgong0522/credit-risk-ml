```python
from pathlib import Path
import sys
import time

# Get project root directory
PROJECT_ROOT = Path(
    "/Users/hongmo/Desktop/Gradient Boosting"
)

# Add project root to Python module search path
sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from src.data_loader import load_and_split_data
from src.preprocessing import build_preprocessor
from src.models import build_baseline_pipeline

# Construct the correct data path
data_path = PROJECT_ROOT / "data" / "application_train.csv"

# Load and split data
X_train, X_val, X_test, y_train, y_val, y_test = load_and_split_data(
    data_path
)

# Create preprocessing pipeline
baseline_preprocessor = build_preprocessor(
    X_train,
    scale_numeric=True
)

# Create baseline model pipeline
baseline_pipeline = build_baseline_pipeline(
    baseline_preprocessor
)

# Train baseline model
start_time = time.perf_counter()

baseline_pipeline.fit(X_train, y_train)

training_time = time.perf_counter() - start_time
print(f"Baseline Training Time: {training_time:.2f} seconds")
```

Calculate Baseline Validation ROC-AUC

```python
from sklearn.metrics import roc_auc_score

# Predict probability of repayment difficulties
y_val_prob = baseline_pipeline.predict_proba(X_val)[:, 1]

# Calculate validation ROC-AUC
val_auc = roc_auc_score(y_val, y_val_prob)

print(f"Baseline Validation ROC-AUC: {val_auc:.4f}")
```

```python
# Training probabilities
y_train_prob = baseline_pipeline.predict_proba(X_train)[:, 1]

# Training ROC-AUC
train_auc = roc_auc_score(y_train, y_train_prob)

print(f"Training AUC:   {train_auc:.4f}")
print(f"Validation AUC: {val_auc:.4f}")
print(f"AUC Gap:        {train_auc - val_auc:.4f}")
```
